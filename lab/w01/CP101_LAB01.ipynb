## 1. What are `.py` and `.ipynb` Files?
Python code lives in one of two common file formats:
- **`.py` files** are plain-text Python scripts: just code, run top-to-bottom as a single program each time, with no embedded output saved in the file.
- **`.ipynb` files** ("IPython Notebook") are structured documents made of separate cells (markdown text cells and code cells) that can be run individually and in any order against a shared, persistent kernel, with each code cell's output saved back into the file.

This class works almost entirely in `.ipynb` notebooks, since mixing explanation, code, and output in one place is useful for data analysis and for showing your work.

## 4. Working with Text and Markdown Cells
### What is a Text Cell?
A text cell contains Markdown-formatted documentation that renders as formatted text when executed, used for explanations and instructions.

### Understanding Markdown Used in Text Cells
**Markdown** is a lightweight markup language that allows you to format text using simple syntax. It's widely used for:
* Documentation
* README files
* Jupyter notebook text cells
* Websites and blogs

<br>

#### Some helpful examples of markdown syntax:
Note that this syntax might vary slightly from platform to platform

<br>

##### Formatting Headers

```markdown
    # Header 1 (Largest)
    ## Header 2
    ### Header 3
    #### Header 4
    ##### Header 5
    ###### Header 6 (Smallest)
```

##### Formatting Text

```markdown
    **Bold text**
    *Italic text*
    ***Bold and italic***
    ~~Strikethrough~~
```

##### Formatting Lists

```markdown
    - Unordered list item 1
    - Unordered list item 2
    - Nested item

    1. Ordered list item 1
    2. Ordered list item 2
      1. Nested ordered item
```

##### Links and Images

```markdown
    [Link text](URL)
    ![Image alt text](image_URL)
```

##### Code

````markdown
`Inline code`

```python
# Code block
print("Hello, World!")

```
````

##### Tables

```markdown
| Column 1 | Column 2 | Column 3 |
|----------|----------|----------|
| Row 1    | Data     | More data|
| Row 2    | Data     | More data|
```

## 5. Working with Code Cells
### Understanding Code Cells

- Interactive execution: Run individual cells independently
- Output display: Results appear directly below the cell
- State persistence: Variables remain in memory between cell executions
- Execution order: Cells can be run in any order (though this can cause confusion)

#### More on Execution Order
Jupyter notebooks maintain a kernel state that persists across cell executions.

**This means that...**
- Variables defined in one cell are available in other cells
- Cells can be executed in any order
- The number in brackets [1] shows execution order

## 6. Exploring the JupyterLab Interface

### Touring the Interface Elements

#### Menu Bar
| Menu | Functions |
|------|-----------|
| **File** | Create, open, save, rename, download notebooks |
| **Edit** | Cut, copy, paste, find and replace |
| **View** | Toggle panels (file browser, table of contents, etc.) |
| **Run** | Run one cell, run all cells, run all above/below |
| **Kernel** | Restart, interrupt, or change the Python kernel |
| **Tabs** | Switch between open files and panels |
| **Settings** | Theme, keyboard shortcuts, other preferences |

#### Toolbar
| Button | Function |
|--------|----------|
| **Save icon** | Save the notebook |
| **+** | Insert a new cell below |
| **Scissors icon** | Cut the selected cell |
| **▶** | Run the current cell |
| **■** | Interrupt the kernel |
| **↻** | Restart the kernel |
| **Cell-type dropdown** | Switch a cell between Code / Markdown / Raw |

#### Cell Operations & Shortcuts
*(These apply in "command mode" — press `Esc` first, or click the cell's left margin, so the cell border is highlighted rather than the cursor blinking inside it.)*

| Action | Shortcut | Description |
|--------|----------|-------------|
| **Run cell** | `Shift + Enter` | Run and move to next cell |
| **Run cell in place** | `Ctrl + Enter` | Run but stay in current cell |
| **Add cell above** | `A` | Insert new cell above current |
| **Add cell below** | `B` | Insert new cell below current |
| **Delete cell** | `D`, `D` (press twice) | Delete current cell |
| **Change to code** | `Y` | Convert cell to code cell |
| **Change to markdown** | `M` | Convert cell to text/markdown cell |
| **Enter edit mode** | `Enter` | Start typing inside the cell |

#### Left Sidebar Panels
| Panel | Purpose |
|-------|---------|
| **File Browser** | Browse and manage your DataHub files |
| **Running Terminals and Kernels** | See and shut down active sessions |
| **Table of Contents** | Navigate through markdown headers |
| **Extension Manager** | Browse/install JupyterLab extensions (if enabled) |

#### Cell Types
| Type | Description |
|------|-------------|
| **Code cells** | Execute Python code |
| **Markdown cells** | Formatted text/documentation |
| **Raw cells** | Passed through unprocessed (rarely used in this course) |

## 7. Introducing the File Directory and Terminal

#### What is a file directory?
A file directory (or folder) is a way of organizing files in a hierarchical structure on a computer system. Think of it like a filing cabinet:

- Root directory: The main folder (like the filing cabinet itself)
- Subdirectories: Folders within folders (like drawers and folders within drawers)
- Files: Individual documents stored in directories

#### DataHub's File System
On DataHub, you have access to:
  1. Your Home Directory: your personal, persistent storage — everything you save here stays put between sessions, just like a small personal server
  2. Course Materials: any files provided by the course (e.g. a cloned repo or shared folder)
  3. Uploaded Files: files you upload during the session via the file browser

  > *Try opening up your file directory folder.  What do you see?*

#### What is a Terminal?
A terminal (also called command line, command prompt, or shell) is a text-based interface for interacting with your computer's operating system. Instead of clicking icons and menus, you type commands.

##### Common Terminal Commands
This lab only needs the four commands below -- you will use every one of them in the Try It Out that follows. (There are many more shell commands out there, but no need to front-load them all before you have used even these four.)

- `pwd`: Print working directory (show current location)
- **`ls`: List files and directories**
- **`cd`: Change directory**
- **`mkdir`: Create new directory**

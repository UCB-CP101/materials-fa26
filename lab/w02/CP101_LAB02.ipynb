## What is Computational Thinking?

Before we write a single line of code, we need to learn how to **think computationally**. This doesn't mean becoming a robot—it means developing a systematic approach to problem-solving that computers can help us execute.

Computational thinking has four key components:

| Component | What it means | Urban Planning Example |
|-----------|--------------|------------------------|
| **Decomposition** | Breaking a complex problem into smaller, manageable parts | Breaking "improve transit access" into: identify underserved areas, map existing routes, analyze demographics, propose new stops |
| **Pattern Recognition** | Finding similarities or trends in data | Noticing that traffic accidents cluster at intersections without protected left turns |
| **Abstraction** | Focusing on important information while ignoring irrelevant details | Representing a city as a network of nodes (intersections) and edges (streets) rather than every building |
| **Algorithm Design** | Creating step-by-step instructions to solve a problem | A process for determining which potholes to fix first based on severity, traffic volume, and budget |

Should we include ALL of this information in our analysis?

### The Power of Abstraction

**Abstraction** means keeping only what matters for our specific question and ignoring the rest.

For bus route planning, we might abstract Oakland's streets into:
- **Nodes**: Intersections and bus stops
- **Edges**: Street segments connecting nodes
- **Attributes**: Travel time, population density along segment

We deliberately **ignore**: building colors, tree species, fire hydrant locations—these don't affect bus routing.

### The Scenario

You work for the Oakland Department of Transportation. You have:
- 500 reported potholes
- Budget to fix 100 potholes this month
- Data on each pothole: size, location, traffic volume, time since reported

### What is an Algorithm?

An **algorithm** is a step-by-step procedure for solving a problem. It's like a recipe: if you follow the steps exactly, you'll get the same result every time.

Algorithms must be:
- **Unambiguous**: Each step is clear
- **Finite**: It eventually ends
- **Effective**: Each step can actually be performed

### Example Algorithm: Making a PB&J Sandwich

```
1. Get two slices of bread
2. Get peanut butter jar and jelly jar
3. Get a knife
4. Open peanut butter jar
5. Use knife to scoop peanut butter
6. Spread peanut butter on one slice of bread
7. Open jelly jar
8. Use knife to scoop jelly
9. Spread jelly on other slice of bread
10. Press the two slices together, spreads facing inward
11. Done!
```

## Activity 5: Putting It All Together - The Complete Workflow

### The Big Picture

In this course, you'll learn to solve urban planning problems using this workflow:

```
Real-World Problem
       ↓
   [DECOMPOSITION]
       ↓
Smaller Sub-Problems
       ↓
   [PATTERN RECOGNITION]
       ↓
Similarities / Reusable Structures
       ↓
   [ABSTRACTION]
       ↓
Essential Details + Model (ignore irrelevant detail)
       ↓
   [ALGORITHM DESIGN]
       ↓
Step-by-Step Solution
       ↓
   [EVALUATION / ITERATION]
       ↓
Test, Debug, Refine (repeat as needed)

```

## Reflection: Why Computational Thinking Before Coding?

You might be eager to start writing Python code. So why did we spend this whole lab on computational thinking?

Here's the secret: **The hard part of programming isn't the code Syntax, it's the thinking.**

Professional programmers spend most of their time:
- Understanding the problem
- Breaking it into pieces
- Designing solutions
- Debugging their thinking

The actual typing? That's maybe 20% of the job.

When we start Python next week, you'll see how these computational thinking skills translate directly into code. Loops implement algorithms. Functions encapsulate sub-problems. Data structures are abstractions. Pattern recognition becomes data analysis.

**The better you are at computational thinking, the better programmer you'll become.**

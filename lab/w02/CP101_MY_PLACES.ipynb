## A note on commas

A CSV separates values with commas. Addresses **contain** commas. So this line is wrong:

```text
Berkeley Bowl,2020 Oregon St, Berkeley, CA 94703,37.8557,-122.2725,food,,,
```

It looks fine, but the computer reads ten values instead of eight, because every comma counts as a separator.

**Wrap any value containing a comma in double quotes:**

```text
Berkeley Bowl,"2020 Oregon St, Berkeley, CA 94703",37.8557,-122.2725,food,,,
```

Now it is eight values, and the quotes tell the computer to treat the whole address as one.

### A complete example

```text
name,address,lat,lon,category,notes,rating,accessibility
Free Speech Movement Cafe,"3 Moffitt Library, Berkeley, CA 94720",37.8725,-122.2606,food,open late during finals,love it,
Doe Memorial Library,"Doe Library, University Dr, Berkeley, CA 94720",37.8722,-122.2595,reading,north reading room is the quiet one,,
```

Both rows end with a comma because `accessibility` is empty. The second row ends with two, because it leaves `rating` empty as well. Blank values are fine; the commas still have to be there, since that is how the computer counts which value belongs to which column.

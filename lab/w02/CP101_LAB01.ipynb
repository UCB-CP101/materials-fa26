Labs that read data carry one more thing, a `data/` folder with a `raw/` folder inside it.
This lab has no data of its own, but you will meet that layout from Lab 3 onward, and it is
worth knowing what it means now.

### Why `data/raw/`

Nearly every lab in this course reads external data from a `data/raw/` folder, and the convention is worth naming because it is a professional habit rather than a Jupyter rule.

**`raw` means untouched.** Whatever is in there is the data as it came from the Census Bureau, the city portal, or the survey. You read from it and you never write back to it. Your cleaning happens in memory, in the notebook, where anyone can see exactly what you did to the data.

The reason is reproducibility. If you clean a file in place, the cleaning becomes invisible and permanent: nobody else can rerun your work from the source, including you in three weeks. Keeping the raw file untouched means your notebook is a complete, checkable account of how you got from the source to your result.

### Paths are relative

When a lab reads a file, it does so like this:

```python
pd.read_csv('data/raw/some_file.csv')
```

That is a **relative path**: it means "starting from the folder this notebook is in". Written that way, the code runs on your DataHub account, on your GSI's, and on the autograder, because all three have the notebook and its `data/` folder sitting together. An **absolute path** like `/home/jovyan/CP101/data/raw/some_file.csv` works only on the one machine it was typed on, which is why you will lose points for them later in the course.

## 1.5 When something goes wrong

Work through these in order. The first one fixes most things.

| Symptom | What to try |
|---|---|
| A cell runs forever, or the notebook stops responding | **Kernel → Interrupt Kernel**, then **Kernel → Restart Kernel** |

**Restarting the kernel does not delete anything.** It throws away the variables Python is holding in memory and gives you a clean slate; your notebook, your code, and your files are untouched. When in doubt, restart.

## 2.1 `.py` files and `.ipynb` files

Python code lives in one of two formats, and you just looked at one of each.

- **`.py` files** are plain-text scripts: just code, run top to bottom as a single program each time, with no output stored in the file. `sample_py_file.py` is one.
- **`.ipynb` files** ("IPython Notebook") are structured documents made of separate **cells**. Cells can be run individually and in any order, they share one live Python session, and each code cell's output is saved back into the file underneath the code that produced it. This notebook is one.

This course works almost entirely in notebooks, because mixing explanation, code, and results in one document is exactly what data analysis needs. When you get to writing reusable functions and scripts, `.py` files come back.

| | `.py` | `.ipynb` |
|---|---|---|
| Structure | one continuous file | separate cells |
| Execution | top to bottom, every time | any cell, any order |
| Output | printed to a terminal, not stored | saved into the file |
| Narration | `#` comments only | full markdown text cells |
| Good for | reusable programs, scripts | analysis, teaching, showing your work |

## 2.3 Command mode and edit mode

This is the piece that trips everyone up, so it comes before the shortcuts rather than after.

JupyterLab has two modes:

- **Edit mode**: there is a cursor inside the cell and typing types. Get there by pressing `Enter` or clicking into the cell.
- **Command mode**: the cell is selected as a whole and single keystrokes are commands. Get there by pressing `Esc` or clicking the cell's left margin. The border is highlighted and there is no blinking cursor.

**Every keyboard shortcut below is a command mode shortcut.** If pressing `B` types the letter b into your cell, you are in edit mode. Press `Esc` first.

| Action | Key | Note |
|---|---|---|
| Insert cell **below** | `B` | |
| Insert cell **above** | `A` | |
| Change to **markdown** | `M` | |
| Change to **code** | `Y` | |
| **Delete** cell | `D`, `D` | press it twice |
| **Undo** cell operation | `Z` | brings a deleted cell back |
| Run and move on | `Shift+Enter` | |
| Run and stay put | `Ctrl+Enter` | |

> `Z` in command mode undoes **cell** operations. `Ctrl+Z` in edit mode undoes your **typing**. After a stray `D`,`D` the one you want is `Z`.

## 2.4 Running cells, and the kernel

Behind your notebook is a **kernel**: a live Python session that stays running between cells. This is the single most important thing to understand about notebooks.

Because the kernel persists:

- A variable defined in one cell is available in every other cell.
- Cells can be run in **any order**, and the kernel remembers whatever you last told it.
- The `[1]`, `[2]` numbers in the left margin are the **execution count**: the order cells were actually run in, which is not necessarily the order they appear on the page.
- A cell showing `[*]` is still running.

### Why this bites people

Suppose you define `x = 5`, run the cell, then edit it to `x = 500` but do not rerun it. The kernel still holds `5`. Your notebook now says one thing on screen and another in memory, and every result below it is computed from a value that is not written anywhere.

The habit that prevents this: before you submit anything, **Kernel → Restart Kernel and Run All Cells**. If it runs top to bottom without error, your notebook actually says what it appears to say. If it does not, you have found a real problem and you found it before your GSI did.

You will see this instruction in every lab from here on. This is why.

## 2.6 How your work gets checked

Most labs in this course, including this one, are graded by a computer. Here is how that works, so nothing about it is a surprise.

### The checking happens after you submit

When you upload your zip to Gradescope, an **autograder** opens your notebook, runs it, and tests specific answers. A few moments later Gradescope shows you the result: which checks passed, which did not, and a message for each one that failed.

So you do not have to guess whether you got it right. You submit, and you are told.

When a check fails you do not get a bare `AssertionError`. You get a message written to tell you what went wrong:

```text
AssertionError: twenty_twenty_six should work out to 2026. If this failed, check the
arithmetic in your expression.
```

Read those messages. They are the fastest help you will get in this course, and they are written by the same people who wrote the question.

**You can resubmit** as many times as you like before the deadline, so a failed check is information, not a verdict. Fix it, export again, upload again.

In later labs you will meet cells like `grader.check("q1")`, which run a question's tests right there in the notebook so you can check yourself before submitting. This lab does not have them, on purpose: today the point is to see the submit-and-get-graded loop from end to end.

During this course, you have spent dozens of hours learning to clean, analyze and organize data, then create visualizations using it. 

But __why?__ How will these visualizations be used in practice? In what context will they appear? And what do they actually communicate about the urban environment?

While a great visualization should be understandable on its own, the broader takeaways can be difficult to communicate without context. Moreover, some visualizations may make more sense in conversation with one another. 

In other words, __the context invariably impacts your interpretation__ of a visualization.

Ultimately, we want to use our data analysis to answer a compelling question about the urban environment. But doing the work to answer that question is only half the battle. We then have to communicate our findings, as well as the question that inspired us to seek them. 

To do so, we need to consider our audience. While we are producing our work in an academic setting, you will also have the ability to make your work publicly available and potentially share it with a non-technical audience. 

Think about someone else besides the teaching team reading your report. How might your orient the reader to your subject quickly? How might you get the reader’s attention, and persuade them not to click away?

# On using AI in your writing

##### Don't.

Writing is the __most direct expression of your thoughts__. Using AI tools to write __abstracts you from your own thinking__ and produces a product that is inherently __impersonal__, both to the reader __and to yourself__. 

Working through the writing process will help you work through your thoughts and your analysis. 

Especially if you are using free AI tools, __generated writing is not fun or interesting to read__. It often produces work that seems very thoughtful and clear at first, but actually __communicates very little__ and feels __devoid of meaning__. 

Moreover, it locks you into a specific frame of mind. Once you see the generated version, it will be hard for you to think outside the box, or create something that __feels distinct to you__.   

In a world where __everyone has access to AI__, the best thing you can do to stand out is to develop __your own voice and style__.

Lastly, you are here to learn. If you outsource that learning to AI, you should ask yourself: __why are you here?__

While this course allows for the occasional use of AI for your coding (as long as you still understand and can explain what every line of code does), it is not allowed for other uses.

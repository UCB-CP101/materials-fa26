## Summary: Plotting and Mapping Census Data

| Step | Takeaway |
|---|---|
| `sns.histplot()` / `sns.scatterplot()` | Quick distribution and relationship plots for a numeric census variable |
| ACS sentinel value `-666666666` | The Census Bureau's marker for a suppressed or unreliable estimate, not a real value |
| Filtering `> 0` on an income column | Removes suppressed rows before they distort a plot's scale |
| `GEOID` built from `state` + `county` + `tract` | The standard key for joining Census tabular data to TIGER/Line geometries |
| `GeoDataFrame.explore()` | Produces an interactive Folium map colored by a chosen column |

## Key Takeaways

1. **Always inspect summary statistics (`.describe()`) before trusting a plot** - a single extreme value can distort a histogram or scatterplot's scale.
2. **ACS data uses sentinel values like `-666666666`** to mark suppressed estimates; filter these out before analysis.
3. **A `GEOID` built from `state`, `county`, and `tract` codes** is the standard way to join Census attribute tables to their geometries.
4. **`.explore()` turns a `GeoDataFrame` into an interactive, shareable web map** with tooltips and popups in a few lines of code.

## 1. Python as a calculator

Python evaluates arithmetic with the operators you already know, plus a couple that are
especially handy for counting things.

| Operator | Meaning | Example |
|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `4500 + 380` |
| `/` | division (always a decimal) | `9 / 2` -> `4.5` |
| `//` | floor division (drop the remainder) | `9 // 2` -> `4` |
| `%` | modulo (the remainder) | `9 % 2` -> `1` |
| `**` | power | `2 ** 10` |

```python
# weekday entries at two BART stations
downtown_berkeley = 4521
ashby = 2980

downtown_berkeley + ashby        # total riders across the two stations
```

```python
# a decimal result, even when it divides evenly
84 / 2
```

### Order of operations

Python follows the usual math precedence: powers first, then multiplication and division,
then addition and subtraction. Use parentheses when you want a different order.

```python
print(2 + 3 * 4)      # multiplication happens first -> 14
print((2 + 3) * 4)    # parentheses change the order -> 20
```

### Floor division and modulo

These two are a pair, and they show up constantly in planning problems. `//` tells you
*how many whole groups* you can make, and `%` tells you *what is left over*.

```python
waiting = 1547         # passengers on the platform
capacity = 200         # passengers per train

full_trains = waiting // capacity   # whole trains we can fill
left_over   = waiting % capacity     # passengers waiting for the next one
print("full trains:", full_trains)
print("left over:  ", left_over)
```

## 3. Values and their types

Every value in Python has a **type**. The four you will meet first:

- `int` : a whole number, like `4521`
- `float` : a number with a decimal point, like `37.8715`
- `str` : text, in quotes, like `"Ashby"`
- `bool` : a truth value, either `True` or `False`

Ask Python for a type with the `type()` function.

```python
print(type(4521))        # int
print(type(37.8715))     # float
print(type("Ashby"))     # str
print(type(True))        # bool
```

### Converting between types

Sometimes a value has the wrong type for what you want to do, for example a number that
arrived as text. Convert it with `int()`, `float()`, or `str()`.

```python
entries_text = "4521"          # this is a string, not a number
entries = int(entries_text)     # now it is an int
print(entries * 2)              # 9042, real arithmetic

print("The station had " + str(entries) + " entries.")  # number back to text
```

## 4. Strings

A **string** is a piece of text in single or double quotes. `+` joins strings together
(concatenation) and `*` repeats them.

```python
line = "Richmond"
color = "Red"
print(color + " Line to " + line)     # join with spaces of our own
print("-" * 30)                        # a divider, 30 dashes
```

### Indexing and slicing

The characters of a string are numbered from `0`. Use square brackets to pull out one
character, or a **slice** `[start:stop]` to pull out a range (the `stop` index is not
included). Negative numbers count from the end.

```python
station = "MacArthur"
print(station[0])      # 'M', the first character
print(station[-1])     # 'r', the last character
print(station[0:3])    # 'Mac', characters 0, 1, 2
print(len(station))    # 9, how many characters
```

### String methods

A **method** is a function attached to a value, called with a dot. Strings have many for
cleaning and checking text.

```python
raw = "  downtown BERKELEY  "
print(raw.strip())              # remove the spaces at the ends
print(raw.strip().title())      # 'Downtown Berkeley'
print(raw.upper())              # ALL CAPS
print("Berkeley" in raw)        # True: is that substring present?
print(raw.replace("BERKELEY", "Oakland"))
```

### f-strings: putting values inside text

An **f-string** (note the `f` before the quote) lets you drop variables straight into a
string with `{ }`. It is the cleanest way to build a readable line of output.

```python
name = "Ashby"
entries = 2980
print(f"{name} had {entries} entries today.")
print(f"That is about {entries / 7:.0f} per day on average.")   # :.0f = no decimals
```

## 5. Variables

A **variable** is a name that refers to a value. The `=` sign is the **assignment
operator**: it takes the value on the right and binds the name on the left to it. It does
*not* mean "equals" in the math sense.

```python
station_name = "Downtown Berkeley"
daily_entries = 4521
latitude = 37.8698

# variables can be used in expressions, just like the values they hold
annual_estimate = daily_entries * 260     # roughly 260 weekday services a year
print(station_name, "->", annual_estimate)
```

### Naming variables well

Good names make code readable. The rules and the house style:

- start with a letter or `_`, then letters, digits, or `_` (no spaces, no dashes);
- cannot be a Python keyword (`for`, `if`, `class`, ...);
- names are case-sensitive (`Ashby` and `ashby` differ);
- **use `snake_case`**: lowercase words joined by underscores, like `daily_entries`.

A name should say what the value *means*, so `daily_entries` beats `x` or `de`.

```python
# clear, snake_case names
median_income = 96400
has_elevator = True
print(median_income, has_elevator)
```

### Reassignment

A variable can be pointed at a new value at any time. The old value is simply forgotten.

```python
riders = 4521
print(riders)
riders = riders + 100     # update: take the current value, add 100, rebind the name
print(riders)
```

## 8. Printing, and expressions vs statements

`print()` displays its arguments. Give it several, separated by commas, and it puts a
space between them. Two useful keyword arguments: `sep` (what to place between items) and
`end` (what to put at the very end, a newline by default).

```python
print("Ashby", "Rockridge", "MacArthur")               # spaces between
print("Ashby", "Rockridge", "MacArthur", sep=" -> ")   # custom separator
print("loading", end="... ")
print("done")                                          # stays on the same line
```

An **expression** is anything that produces a value (`4521 + 100`, `station.upper()`). A
**statement** is an instruction that performs an action (an assignment, an `import`, a
`print`). A cell can hold several statements; only the value of the last bare expression
is shown automatically, which is why we use `print()` to see more than one result.

```python
x = 5 * 5      # a statement: it acts (binds x), it does not display anything
x              # an expression on its own: its value, 25, is displayed
```

## Glossary

- **value** : a basic piece of data, such as `4521` or `"Ashby"`.
- **type** : the kind of a value (`int`, `float`, `str`, `bool`).
- **expression** : a combination of values and operators that produces a value.
- **statement** : an instruction that performs an action (assignment, `import`, `print`).
- **variable** : a name that refers to a value.
- **assignment** : binding a name to a value with `=`.
- **function** : a named operation you call with arguments in parentheses.
- **method** : a function attached to a value, called with a dot (`text.upper()`).
- **module** : a file of extra tools you bring in with `import`.
- **comment** : text after `#` that Python ignores.

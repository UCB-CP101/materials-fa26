## Static Webpage hosting with Github Pages and Jekyll

The materials are partially adapted from the [Jekyll](https://jekyllrb.com/docs/) and [GitHub Pages](https://docs.github.com/en/pages/setting-up-a-github-pages-site-with-jekyll/about-github-pages-and-jekyll) documentation.

Let's use your new Github Pages project site to generate a static webpage that hosts information about your project.

We will be using Jekyll to build your site. Jekyll is a static site generator where you write your pages in Markdown, and Jekyll converts them into a static HTML page, which is then styled using CSS. The benefit of using Markdown is that you will not need to learn HTML, CSS, and JavaScript. 

Jekyll is built into GitHub Pages, so every time you commit a change to your repository, GitHub runs Jekyll for you and publishes the result. That means there is nothing to install and no build commands to run. 

Building a Jekyll site on Github Pages broadly consists of **four steps**:

1. **Configure your site.** Tell Github Pages where your site lives and give it a title using a `_config.yml` file.

2. **Add content to your site.** Write your pages as Markdown (.md) files.

3. **Build your site.** Github Pages runs Jekyll automatically every time you commit.

4. **View your site online.** Once the build finishes, your site is live at your Github Pages link.

**Step 1: Configure Your Site**
1. In your Github Pages repository on github.com, click "Settings" and then "Pages" in the left sidebar. Under "Build and deployment", set the "Source" to "Deploy from a branch", then pick the `main` branch and the `/ (root)` folder and click "Save". This tells Github Pages to build your site with Jekyll from the files on your `main` branch.

2. Return to the main page of your repo. Click the "+" sign next to the "Code" button, click "Create new file" and name it `_config.yml`. This file is known as the "configuration" file, and it holds settings that apply to your whole site, such as its title and description. Add the following, fill in your own title and description, and click "Commit changes":

```
title: My Project
description: Project Description
```

3. Notice that we did not pick a theme. When a site doesn't name one, Github Pages styles it with its default theme, which is all we need for today. If your team wants a different look later, see the [supported themes](https://pages.github.com/themes/) list.

**Step 2: Add content to your Site**

1. Every Jekyll site needs a homepage. Create a new file called `index.md` in the root of your repo (not inside one of your numbered folders). If your repo already has an `index.md` file open it and click the pencil icon to edit it instead. NOTE: if your repo has a `README.md` but no `index.md`, Jekyll uses the README as your homepage, so creating `index.md` makes it clear which file is your homepage.

Add the following to your file:

```
---
title: Home
---

# Lab 10 GitHub Basics

This is text explaining my project.

Yadda Yadda Yadda.

```
The block between the two `---` lines at the top is called **front matter**. Jekyll reads settings for that page, like the page title, from it. Front matter is optional on Github Pages, but it's good practice to include it on every page. Everything below the front matter is regular Markdown. You can read more about [front matter](https://jekyllrb.com/docs/front-matter/) and [Markdown syntax](https://docs.github.com/en/get-started/writing-on-github/getting-started-with-writing-and-formatting-on-github/basic-writing-and-formatting-syntax).

Commit your changes.

**Step 3: Build your Site**

1. You already did! Each time you clicked "Commit changes", Github Pages started a new Jekyll build.

2. To watch a build, click the "Actions" tab of your repo. You should see a workflow called "pages build and deployment" for each commit. A yellow dot means it is still running, a green check means your site was published, and a red X means the build failed. If a build fails, click on it to read the error. The most common cause is a typo in `_config.yml`.

**Step 4: View your Site**

1. Once the build shows a green check, your site should be viewable at your site link `https://<your_username>.github.io/<your_repo_name>/`. You can also find the link at the top of "Settings" > "Pages".

2. Yay! This is how you create a static page on Github Pages using Jekyll.
